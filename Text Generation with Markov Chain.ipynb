```python
import random
import re
from collections import defaultdict
```

```python
class MarkovChain:
    def __init__(self):
        self.model = defaultdict(list)

    def preprocess(self, text):
        text = text.lower()
        text = re.sub(r'[^a-z\s]', '', text)
        return text.split()

    def train(self, text):
        words = self.preprocess(text)
        for i in range(len(words) - 1):
            self.model[words[i]].append(words[i + 1])

    def generate_text(self, start_word, length=30):
        start_word = start_word.lower()
        current_word = start_word
        generated = [current_word]

        for _ in range(length - 1):
            next_words = self.model.get(current_word)
            if not next_words:
                break
            current_word = random.choice(next_words)
            generated.append(current_word)

        return " ".join(generated)
```

```python
training_text = """
Once upon a time there was a sleepy dragon who loved coffee
The dragon drank coffee and forgot to breathe fire
A cat learned programming and wrote code all night
The programmer laughed when the computer said hello
Robots danced in the rain and sang silly songs
The moon smiled at the stars and told funny stories
A wizard typed spells on a laptop instead of a book
The pizza delivery robot got lost in space
Every morning the coffee machine dreamed of vacation
The AI wrote poetry and made humans laugh
"""
```

```python
markov = MarkovChain()
markov.train(training_text)

print("✅ Model trained successfully!")
print(f"Total unique words learned: {len(markov.model)}")
```

```python
start_word = input("Enter a starting word: ")
length = int(input("Enter number of words to generate: "))

generated_text = markov.generate_text(start_word, length)

print("\n📝 Generated Text:")
print(generated_text)
```
